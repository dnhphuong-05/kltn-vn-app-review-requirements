## 6. Kiểm tra API OpenAI (chạy một lần, local)

Protocol chốt chiến lược C dùng `gpt-4o-mini` với snapshot cố định `gpt-4o-mini-2024-07-18` và temperature 0. Các snapshot cũ có
thể bị nhà cung cấp gỡ bỏ, nên trước Giai đoạn 8 em gọi API đúng một lần với một review mẫu để chắc snapshot này vẫn dùng được, xem
model mà API thực sự trả về, số token và chi phí của một request. Key được đọc từ `.env` (không ghi trong notebook) và được che trong
mọi thông báo lỗi. Giá dùng để ước lượng là giá tham chiếu, em sẽ đối chiếu lại với trang giá của nhà cung cấp vào ngày chạy GĐ8.

```python
from src.utils.llm import check_api, estimate_cost
r = check_api('app hay bị văng khi thanh toán')
if not r['ok']:
    print(f"[LỖI] {r['error']}: {r.get('hint', '')}\n  chi tiết: {r['detail']}")
else:
    print('Model yêu cầu :', r['requested_model'], '| temperature', r['temperature'])
    print('Model trả về  :', r['response_model'], '| khớp snapshot:', r['snapshot_match'])
    print('Trả lời       :', r['answer'])
    print('Token         :', r['prompt_tokens'], 'prompt +', r['completion_tokens'], 'completion')
    print(f"Chi phí 1 request: ${r['cost_usd']:.6f} (giá/1M token: in {r['prices']['input']}, out {r['prices']['output']}; {r['prices']['source']})")
    N = 1550
    naive = estimate_cost(N, r['prompt_tokens'], r['completion_tokens'])
    # prompt GĐ8 thật dài hơn nhiều: định nghĩa nhãn + quy tắc + ví dụ few-shot (giả định ~1.500 token vào, ~20 token ra);
    # tầng 2 chỉ gọi cho phần actionable (~20,5% theo pilot)
    real = estimate_cost(N, 1500, 20, calls_per_review=1 + 0.205)
    print(f"[OK] snapshot {r['response_model']} dùng được. Ước chi phí ~{N:,} review: "
          f"${naive:.4f} nếu prompt ngắn như test; ~${real:.2f} với prompt few-shot giả định ~1.500 token, 2 tầng.")
```
